## 1. 数据科学基础

### 1.1 NumPy (数值Python)
NumPy是Python中科学计算的基础包。它提供了一个高性能的多维数组对象，以及对这些数组进行操作的工具。

```
import numpy as np
print('--- NumPy Advanced ---')
# Broadcasting
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([10, 20, 30])
print(f'Broadcasting example:\n{a + b}')
# Linear Algebra
A = np.array([[1, 2], [3, 4]])
print(f'Inverse of A:\n{np.linalg.inv(A)}')
```

### 1.2 Pandas
Pandas是一个强大的数据分析和操作库。

```
import pandas as pd
print('--- Pandas Advanced ---')
df = pd.DataFrame({
    'group': ['A', 'B', 'A', 'B'],
    'value': [10, 20, 12, 22]
})
# GroupBy
print(f'Grouped by mean:\n{df.groupby("group").mean()}')
# Merging
left = pd.DataFrame({'key': ['K0'], 'A': ['A0']})
right = pd.DataFrame({'key': ['K0'], 'B': ['B0']})
print(f'Merged:\n{pd.merge(left, right, on="key")}')
```

### 1.3 Matplotlib
Matplotlib是一个用于创建静态、动态和交互式可视化的库。

```
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
print('--- Matplotlib 3D Plot ---')
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
x = np.arange(-5, 5, 0.25)
y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))
ax.plot_surface(X, Y, Z, cmap='viridis')
plt.show()
```

## 2. PyTorch 核心概念

### 2.1 Autograd: 自动求导
PyTorch的autograd包提供了对Tensors上所有操作的自动微分。

```
import torch
print('--- Autograd Higher-Order Derivatives ---')
x = torch.tensor(2.0, requires_grad=True)
y = x**3
grad1 = torch.autograd.grad(y, x, create_graph=True)[0]
grad2 = torch.autograd.grad(grad1, x)[0]
print(f'First derivative (3x^2): {grad1}')
print(f'Second derivative (6x): {grad2}')
```

### 2.2 构建神经网络
使用`torch.nn`包来构建神经网络。

```
import torch.nn as nn
print('--- Neural Network with DataLoader ---')
model = nn.Sequential(nn.Linear(10, 64), nn.ReLU(), nn.Linear(64, 2))
print(f'Model Architecture:\n{model}')
# Saving and loading model state
torch.save(model.state_dict(), 'model.pth')
loaded_model = nn.Sequential(nn.Linear(10, 64), nn.ReLU(), nn.Linear(64, 2))
loaded_model.load_state_dict(torch.load('model.pth'))
print('\nModel loaded successfully.')
```

## 3. 深度学习模型

### 3.1 卷积神经网络 (CNN)
CNNs是图像识别任务的理想选择。

```
print('--- Advanced CNN Block ---')
class ConvBlock(nn.Module):
    def __init__(self, in_c, out_c):
        super().__init__()
        self.conv1 = nn.Conv2d(in_c, out_c, 3, padding=1)
        self.bn1 = nn.BatchNorm2d(out_c)
    def forward(self, x):
        return torch.relu(self.bn1(self.conv1(x)))
block = ConvBlock(3, 64)
print(block)
```

### 3.2 循环神经网络 (LSTM/GRU)
LSTMs和GRUs是为处理序列数据而设计的。

```
print('--- LSTM for Sequence Prediction ---')
lstm = nn.LSTM(input_size=10, hidden_size=20, num_layers=2, bidirectional=True, batch_first=True)
input_seq = torch.randn(3, 5, 10) # batch, seq_len, input_size
output, (hidden, cell) = lstm(input_seq)
print(f'Output shape: {output.shape}')
print(f'Hidden state shape: {hidden.shape}')
```
