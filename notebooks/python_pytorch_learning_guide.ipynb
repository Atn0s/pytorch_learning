## 第1部分：Python基础

### 1.1 变量与数据类型
在Python中，变量是存储数据的容器。无需预先声明类型，Python会根据赋的值自动推断。基本数据类型包括整数（int）、浮点数（float）、字符串（str）和布尔值（bool）。

```
name = "学习者"
age = 25
height = 1.75
is_learning = True
print(f'姓名: {name}, 类型: {type(name)}')
print(f'年龄: {age}, 类型: {type(age)}')
print(f'身高: {height}, 类型: {type(height)}')
print(f'在学习: {is_learning}, 类型: {type(is_learning)}')
```

### 1.2 复合数据类型
Python提供了多种复合数据类型，用于组织和存储多个数据项。

**列表 (List)**: 有序、可变，是Python中最常用的数据结构之一。

```
fruits = ["苹果", "香蕉", "橙子"]
fruits.append("葡萄")
print(f'水果列表: {fruits}')
print(f'第一个水果: {fruits[0]}')
```

**元组 (Tuple)**: 有序、不可变，通常用于存储不应改变的数据。

```
coordinates = (10.0, 20.0)
print(f'坐标: {coordinates}')
```

**字典 (Dictionary)**: 键值对集合，无序（在Python 3.7+中为有序），通过键快速查找值。

```
person = {"name": "张三", "age": 30}
person["city"] = "北京"
print(f'个人信息: {person}')
```

**集合 (Set)**: 无序、不重复的元素集合，常用于去重或数学集合运算。

```
unique_numbers = {1, 2, 3, 2, 1}
print(f'唯一数字: {unique_numbers}')
```

### 1.3 控制结构
控制结构用于决定代码的执行流程，包括条件判断和循环。

**条件判断 (if/elif/else)**

```
score = 88
if score >= 90:
    grade = "优秀"
elif score >= 80:
    grade = "良好"
else:
    grade = "及格"
print(f'分数 {score} 的等级是: {grade}')
```

**For循环**

```
total = 0
for i in range(1, 11):
    total += i
print(f'1到10的和是: {total}')
```

**列表推导式**：一种简洁的创建列表的方式。

```
squares = [x**2 for x in range(1, 6)]
print(f'1到5的平方: {squares}')
```

## 第3部分：PyTorch核心

### 3.1 Tensor: PyTorch的基础
Tensor是PyTorch中的核心数据结构，类似于NumPy的数组，但支持GPU加速和自动求导。

```
import torch
print('--- Tensor操作 ---')

x = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
y = torch.rand_like(x) # 创建一个与x形状相同的随机张量
print(f'Tensor x:\n{x}')
print(f'Tensor y:\n{y}')
print(f'x + y:\n{x.add(y)}')

# GPU加速
if torch.cuda.is_available():
    device = torch.device('cuda')
    x_gpu = x.to(device)
    y_gpu = y.to(device)
    z_gpu = x_gpu + y_gpu
    print(f'GPU上的计算结果:\n{z_gpu}')
else:
    print('未检测到CUDA, 使用CPU进行计算')
```

### 3.2 Autograd: 自动求导
PyTorch的`autograd`包提供了对Tensors上所有操作的自动微分，是构建神经网络的核心。

```
import torch
print('--- Autograd 示例 ---')

# requires_grad=True 表示需要计算梯度
x = torch.tensor(3.0, requires_grad=True)
y = x**2
print(f'y = x^2 = {y}')

# 计算梯度
y.backward()
print(f'dy/dx at x=3: {x.grad}') # 理论值: 2x = 6
```

### 3.3 构建神经网络
使用`torch.nn`包来构建、训练和评估神经网络。这是一个典型的训练流程。

```
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
print('--- 神经网络训练流程 ---')

input_size, hidden_size, output_size, batch_size = 10, 32, 2, 64
# 1. 定义模型
model = nn.Sequential(nn.Linear(input_size, hidden_size), nn.ReLU(), nn.Linear(hidden_size, output_size))
# 2. 准备数据
X = torch.randn(100, input_size)
y = torch.randint(0, output_size, (100,))
loader = DataLoader(TensorDataset(X, y), batch_size=batch_size)
# 3. 定义损失和优化器
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
# 4. 训练循环
model.train()
for epoch in range(2):
    for inputs, labels in loader:
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
    print(f'Epoch {epoch+1}, Loss: {loss.item():.4f}')

# 5. 评估
model.eval()
with torch.no_grad():
    correct = (model(X).argmax(dim=1) == y).sum()
    print(f'Accuracy: {correct.item() / len(y) * 100:.2f}%')
```

## 第4部分：高级模型与项目

### 4.1 卷积神经网络 (CNN)
CNN是图像识别任务的理想选择，它们通过卷积核提取空间特征。一个典型的CNN模块包括卷积层（`Conv2d`）、激活函数（`ReLU`）、池化层（`MaxPool2d`）和批归一化（`BatchNorm2d`）。

```
print('--- CNN 核心模块 ---')
import torch.nn as nn

# 定义一个卷积块
conv_block = nn.Sequential(
    nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, padding=1),
    nn.BatchNorm2d(32),
    nn.ReLU(),
    nn.MaxPool2d(2)
)
dummy_image = torch.randn(1, 3, 32, 32) # (batch, channels, height, width)
output = conv_block(dummy_image)
print(f'输入形状: {dummy_image.shape}')
print(f'输出形状: {output.shape}')
```

### 4.2 循环神经网络 (LSTM/GRU)
LSTM和GRU是为处理序列数据（如文本或时间序列）而设计的，它们能有效捕捉长期依赖关系。`batch_first=True`是一个常用参数，它使输入张量的维度为`(batch, seq_length, feature_size)`。

```
print('--- LSTM 核心模块 ---')
import torch.nn as nn

# 定义一个LSTM层
lstm = nn.LSTM(input_size=20, hidden_size=50, num_layers=2, batch_first=True, bidirectional=True)
dummy_sequence = torch.randn(1, 10, 20) # (batch, seq_length, feature_size)
output, (hidden, cell) = lstm(dummy_sequence)
print(f'输入序列形状: {dummy_sequence.shape}')
print(f'输出形状: {output.shape}') # 双向LSTM，输出特征维度是hidden_size的两倍
print(f'隐藏状态形状: {hidden.shape}')
```

### 5.1 CIFAR-10 图像分类
这个项目展示了如何训练一个类ResNet模型，包含数据增强、学习率调度和详细评估。核心步骤包括：
1. **数据增强**: 使用`transforms.Compose`对训练数据进行随机裁剪、翻转等操作，以增加数据多样性。
2. **模型构建**: 定义一个包含残差块（ResidualBlock）的类ResNet模型。
3. **训练与评估**: 使用学习率调度器（`StepLR`）动态调整学习率，并在训练后生成分类报告。

```
print("--- CIFAR-10 项目核心代码 ---")
import torch
import torch.nn as nn
import torchvision.transforms as transforms

# 数据增强
transform_train = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
])
print("数据增强流程已定义。")

# 这是一个简化的模型定义，实际项目中的模型更复杂
class SimpleResNet(nn.Module):
    def __init__(self):
        super(SimpleResNet, self).__init__()
        self.conv1 = nn.Conv2d(3, 16, 3, 1, 1)
        self.bn1 = nn.BatchNorm2d(16)
        self.relu = nn.ReLU()
        self.fc = nn.Linear(16*32*32, 10)
    
    def forward(self, x):
        x = self.relu(self.bn1(self.conv1(x)))
        x = x.view(x.size(0), -1)
        x = self.fc(x)
        return x

model = SimpleResNet()
print("一个简化的类ResNet模型已定义。详细实现请见 05_projects/image_classification/train_cifar10.py")
```

### 5.2 NLP 情感分析
这个项目展示了如何训练一个基于GRU的模型，并进行适当的文本预处理。核心步骤包括：
1. **文本预处理**: 创建一个词汇表（Vocabulary），将文本转换为数字序列，并使用`collate_fn`进行填充。
2. **模型构建**: 定义一个包含词嵌入（`Embedding`）、GRU和全连接层的`SentimentGRU`模型。
3. **训练与推理**: 使用`BCEWithLogitsLoss`作为损失函数，并在推理时使用`sigmoid`将输出转换为概率。

```
print("--- NLP 情感分析项目核心代码 ---")
import torch
import torch.nn as nn

# 这是一个简化的模型定义
class SentimentGRU(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)
        self.gru = nn.GRU(embedding_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, output_dim)

    def forward(self, text):
        embedded = self.embedding(text)
        _, hidden = self.gru(embedded)
        return self.fc(hidden.squeeze(0))

model = SentimentGRU(vocab_size=1000, embedding_dim=50, hidden_dim=128, output_dim=1)
print("一个简化的情感分析模型已定义。详细实现请见 05_projects/nlp_sentiment/sentiment_analysis.py")
```
