### 1.2 条件语句和循环

控制结构是编程的核心，让我们学习如何使用if语句和循环来控制程序流程。

````python
# 1.2 条件语句和循环
print("=== 条件语句 ===")

# if-elif-else 语句
score = 85
if score >= 90:
    grade = "优秀"
elif score >= 80:
    grade = "良好"
elif score >= 70:
    grade = "中等"
elif score >= 60:
    grade = "及格"
else:
    grade = "不及格"

print(f"分数: {score}, 等级: {grade}")

# for循环遍历列表
print(f"\n=== For循环 ===")
fruits = ["苹果", "香蕉", "橙子", "葡萄"]
print("我喜欢的水果:")
for i, fruit in enumerate(fruits, 1):
    print(f"{i}. {fruit}")

# 使用range()
print(f"\n计算1到10的和:")
total = 0
for i in range(1, 11):
    total += i
print(f"1 + 2 + ... + 10 = {total}")

# while循环
print(f"\n=== While循环 ===")
count = 5
print("倒计时:")
while count > 0:
    print(f"{count}...")
    count -= 1
print("发射！🚀")

# 列表推导式（Python的特色功能）
print(f"\n=== 列表推导式 ===")
numbers = [1, 2, 3, 4, 5]
squares = [x**2 for x in numbers]
even_squares = [x**2 for x in numbers if x % 2 == 0]

print(f"原数字: {numbers}")
print(f"平方: {squares}")
print(f"偶数的平方: {even_squares}")
````

## 3. 函数和模块 🔧

函数是代码复用的基础，模块让我们能够组织和共享代码。

````python
# 3. 函数定义和使用
print("=== 函数基础 ===")

def greet(name, age=None):
    """
    问候函数
    参数:
        name: 姓名 (必需)
        age: 年龄 (可选)
    """
    if age:
        return f"你好，{name}！你今年{age}岁。"
    else:
        return f"你好，{name}！"

# 函数调用
print(greet("小明"))
print(greet("小红", 25))

# 数学函数示例
def calculate_circle_area(radius):
    """计算圆的面积"""
    import math
    return math.pi * radius ** 2

def factorial(n):
    """计算阶乘"""
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(f"\n=== 数学函数示例 ===")
print(f"半径为5的圆面积: {calculate_circle_area(5):.2f}")
print(f"5的阶乘: {factorial(5)}")

# 高阶函数示例
def apply_operation(numbers, operation):
    """对数字列表应用操作"""
    return [operation(num) for num in numbers]

def square(x):
    return x ** 2

def cube(x):
    return x ** 3

numbers = [1, 2, 3, 4, 5]
print(f"\n=== 高阶函数 ===")
print(f"原数字: {numbers}")
print(f"平方: {apply_operation(numbers, square)}")
print(f"立方: {apply_operation(numbers, cube)}")

# Lambda函数（匿名函数）
print(f"使用lambda求平方: {apply_operation(numbers, lambda x: x**2)}")

# 模块导入示例
print(f"\n=== 模块使用 ===")
import random
import datetime

print(f"随机数 (1-10): {random.randint(1, 10)}")
print(f"当前时间: {datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# 从模块导入特定函数
from math import sqrt, sin, pi
print(f"√16 = {sqrt(16)}")
print(f"sin(π/2) = {sin(pi/2):.2f}")
````

## 10. 自动梯度计算 (Autograd) 🎯

自动微分是深度学习的核心！PyTorch的autograd系统能够自动计算梯度，这是训练神经网络的基础。

````python
# 10. 自动梯度计算演示
print("=== 自动梯度基础 ===")

# 创建需要梯度的张量
x = torch.tensor(2.0, requires_grad=True)
print(f"输入 x = {x}")
print(f"x.requires_grad = {x.requires_grad}")

# 定义函数 y = x^2 + 3x + 1
y = x**2 + 3*x + 1
print(f"y = x² + 3x + 1 = {y}")

# 计算梯度 dy/dx = 2x + 3
y.backward()  # 反向传播
print(f"dy/dx = {x.grad}")
print(f"理论值：2*{x.data} + 3 = {2*x.data + 3}")

print(f"\n=== 多变量梯度 ===")
# 重置梯度
x = torch.tensor(1.0, requires_grad=True)
y = torch.tensor(2.0, requires_grad=True)

# 函数 z = x²y + y³
z = x**2 * y + y**3
print(f"x = {x.data}, y = {y.data}")
print(f"z = x²y + y³ = {z}")

z.backward()
print(f"∂z/∂x = {x.grad} (理论值: 2xy = {2*x.data*y.data})")
print(f"∂z/∂y = {y.grad} (理论值: x² + 3y² = {x.data**2 + 3*y.data**2})")

print(f"\n=== 梯度在神经网络中的应用 ===")
# 模拟简单的线性回归
# 数据：y = 2x + 1 + noise
torch.manual_seed(42)  # 保证结果可重复
x_data = torch.randn(100, 1)
y_true = 2 * x_data + 1 + 0.1 * torch.randn(100, 1)

# 模型参数（需要学习）
w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

print(f"真实参数: w=2, b=1")
print(f"初始参数: w={w.data:.3f}, b={b.data:.3f}")

# 训练几步
learning_rate = 0.01
for epoch in range(5):
    # 前向传播
    y_pred = w * x_data + b
    
    # 计算损失（均方误差）
    loss = torch.mean((y_pred - y_true)**2)
    
    # 反向传播
    loss.backward()
    
    # 更新参数
    with torch.no_grad():  # 更新参数时不计算梯度
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad
        
        # 清零梯度
        w.grad.zero_()
        b.grad.zero_()
    
    print(f"Epoch {epoch+1}: Loss={loss.item():.4f}, w={w.data:.3f}, b={b.data:.3f}")

print(f"\n=== 计算图和梯度流 ===")
# 演示复杂计算图
x = torch.tensor(1.0, requires_grad=True)
a = x + 2
b = a * 3
c = b ** 2
d = torch.sin(c)

print(f"计算图: x -> (+2) -> a -> (*3) -> b -> (^2) -> c -> sin() -> d")
print(f"x={x.data:.3f}, a={a.data:.3f}, b={b.data:.3f}, c={c.data:.3f}, d={d.data:.3f}")

d.backward()
print(f"dd/dx = {x.grad:.3f}")

print(f"\n=== 梯度控制 ===")
# 有时我们需要控制梯度计算
x = torch.tensor(2.0, requires_grad=True)

# 1. 停止梯度传播
with torch.no_grad():
    y = x ** 2  # 这里不会计算梯度
    print(f"no_grad下计算: y = {y}, requires_grad = {y.requires_grad}")

# 2. 分离张量（detach）
y = x ** 2
z = y.detach()  # z不会传播梯度到x
print(f"detach后: z.requires_grad = {z.requires_grad}")

# 3. 临时禁用梯度
x.requires_grad_(False)  # 就地修改
print(f"禁用梯度后: x.requires_grad = {x.requires_grad}")

print(f"\n=== 实用技巧 ===")
print("""
梯度计算的重要概念：
1. requires_grad=True: 告诉PyTorch追踪这个张量的梯度
2. .backward(): 计算梯度并累积到.grad属性中
3. .grad.zero_(): 清零梯度（重要！梯度会累积）
4. with torch.no_grad(): 临时禁用梯度计算
5. .detach(): 从计算图中分离张量

在深度学习中：
- 模型参数需要梯度：requires_grad=True
- 输入数据通常不需要梯度
- 损失函数对参数的梯度用于更新参数
""")
````

## 11. 构建神经网络 🧠

现在我们来学习如何使用PyTorch构建神经网络！我们将使用`torch.nn`模块来定义网络结构。

````python
# 11. 构建神经网络
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F

print("=== 简单的线性网络 ===")

# 定义一个简单的神经网络类
class SimpleNet(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SimpleNet, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)  # 第一个全连接层
        self.fc2 = nn.Linear(hidden_size, output_size) # 第二个全连接层
    
    def forward(self, x):
        x = torch.relu(self.fc1(x))  # 隐藏层 + ReLU激活
        x = self.fc2(x)              # 输出层
        return x

# 创建网络实例
net = SimpleNet(input_size=4, hidden_size=10, output_size=3)
print(f"网络结构:\n{net}")

# 查看网络参数
print(f"\n=== 网络参数 ===")
total_params = 0
for name, param in net.named_parameters():
    print(f"{name}: {param.shape}")
    total_params += param.numel()
print(f"总参数数量: {total_params}")

# 测试前向传播
print(f"\n=== 前向传播测试 ===")
dummy_input = torch.randn(1, 4)  # 批次大小为1，特征维度为4
output = net(dummy_input)
print(f"输入形状: {dummy_input.shape}")
print(f"输出形状: {output.shape}")
print(f"输出值: {output}")

print(f"\n=== 使用Sequential构建网络 ===")
# 另一种构建网络的方式
sequential_net = nn.Sequential(
    nn.Linear(4, 10),
    nn.ReLU(),
    nn.Linear(10, 3)
)
print(f"Sequential网络:\n{sequential_net}")

print(f"\n=== 常用层介绍 ===")
# 展示各种常用层
print("线性层 (全连接层):")
linear = nn.Linear(5, 3)
print(f"  输入5维 -> 输出3维: {linear}")

print("卷积层 (用于图像处理):")
conv = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3)
print(f"  3通道 -> 16通道，3x3卷积核: {conv}")

print("激活函数:")
print(f"  ReLU: {nn.ReLU()}")
print(f"  Sigmoid: {nn.Sigmoid()}")
print(f"  Tanh: {nn.Tanh()}")

print("标准化层:")
print(f"  BatchNorm1d: {nn.BatchNorm1d(10)}")

print("Dropout层 (防止过拟合):")
print(f"  Dropout: {nn.Dropout(0.5)}")

print(f"\n=== 损失函数 ===")
# 常用损失函数
print("分类任务 - 交叉熵损失:")
criterion_cls = nn.CrossEntropyLoss()
print(f"  {criterion_cls}")

print("回归任务 - 均方误差:")
criterion_reg = nn.MSELoss()
print(f"  {criterion_reg}")

# 演示损失计算
pred = torch.randn(3, 5)  # 3个样本，5个类别的预测
target = torch.tensor([1, 3, 2])  # 真实标签
loss = criterion_cls(pred, target)
print(f"分类损失示例: {loss.item():.4f}")

print(f"\n=== 优化器 ===")
# 创建优化器
optimizer = optim.Adam(net.parameters(), lr=0.001)
print(f"Adam优化器: {optimizer}")

# 其他常用优化器
sgd_optimizer = optim.SGD(net.parameters(), lr=0.01, momentum=0.9)
print(f"SGD优化器: {sgd_optimizer}")

print(f"\n=== 完整的训练步骤演示 ===")
# 生成一些模拟数据
torch.manual_seed(42)
X = torch.randn(100, 4)  # 100个样本，4个特征
y = torch.randint(0, 3, (100,))  # 100个标签，3个类别

# 创建网络、损失函数和优化器
model = SimpleNet(4, 10, 3)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

print("开始训练...")
for epoch in range(5):
    # 前向传播
    outputs = model(X)
    loss = criterion(outputs, y)
    
    # 反向传播
    optimizer.zero_grad()  # 清零梯度
    loss.backward()        # 计算梯度
    optimizer.step()       # 更新参数
    
    # 计算准确率
    _, predicted = torch.max(outputs.data, 1)
    accuracy = (predicted == y).float().mean()
    
    print(f"Epoch {epoch+1}/5, Loss: {loss.item():.4f}, Accuracy: {accuracy:.4f}")

print(f"\n=== 模型评估 ===")
model.eval()  # 切换到评估模式
with torch.no_grad():  # 评估时不需要计算梯度
    test_outputs = model(X[:10])  # 测试前10个样本
    _, test_predicted = torch.max(test_outputs, 1)
    test_actual = y[:10]
    
    print("预测 vs 实际:")
    for i in range(10):
        print(f"样本{i+1}: 预测={test_predicted[i].item()}, 实际={test_actual[i].item()}")

print(f"\n=== 保存和加载模型 ===")
# 保存模型
torch.save(model.state_dict(), 'model.pth')
print("模型已保存到 model.pth")

# 加载模型
new_model = SimpleNet(4, 10, 3)
new_model.load_state_dict(torch.load('model.pth'))
print("模型已从 model.pth 加载")

print(f"\n=== 学习要点总结 ===")
print("""
神经网络构建的关键步骤：

1. 定义网络结构:
   - 继承 nn.Module
   - 在 __init__ 中定义层
   - 在 forward 中定义前向传播

2. 训练循环:
   - 前向传播: outputs = model(inputs)
   - 计算损失: loss = criterion(outputs, targets)
   - 反向传播: loss.backward()
   - 更新参数: optimizer.step()
   - 清零梯度: optimizer.zero_grad()

3. 重要概念:
   - model.train() / model.eval(): 训练/评估模式
   - torch.no_grad(): 禁用梯度计算
   - 正确的数据类型和形状很重要！

继续探索更复杂的网络结构和应用吧！
""")
````
