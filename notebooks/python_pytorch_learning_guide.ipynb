## 第1部分：数据科学基础

### 1.1 NumPy (数值Python)
NumPy是Python中科学计算的基础包。它提供了一个高性能的多维数组对象，以及对这些数组进行操作的工具。

```
import numpy as np
print('--- NumPy 核心功能 ---')

# 创建数组
a = np.array([1, 2, 3])
print(f'一维数组: {a}')
b = np.zeros((2,3))
print(f'全零数组:\n{b}')

# 数组运算
c = np.ones((2, 3)) * 5
print(f'数组c:\n{c}')
d = c + b
print(f'c + b:\n{d}')

# 广播 (Broadcasting)
e = np.array([10, 20, 30])
print(f'广播示例 d + e:\n{d + e}')

# 线性代数
A = np.array([[1, 2], [3, 4]])
print(f'矩阵A的逆:\n{np.linalg.inv(A)}')
```

### 1.2 Pandas
Pandas是一个强大的数据分析和操作库，它基于NumPy构建，提供了更高级的数据结构和分析工具。

```
import pandas as pd
print('--- Pandas 核心功能 ---')

# 创建DataFrame
data = {'产品': ['苹果', '香蕉', '苹果', '香蕉'], '销量': [100, 150, 120, 130], '城市': ['北京', '上海', '北京', '上海']}
df = pd.DataFrame(data)
print(f'原始DataFrame:\n{df}')

# 分组 (GroupBy)
print(f'按产品分组计算平均销量:\n{df.groupby("产品")["销量"].mean()}')

# 数据透视表
pivot = df.pivot_table(values='销量', index='产品', columns='城市')
print(f'数据透视表 (产品-城市-销量):\n{pivot}')
```

Matplotlib是一个用于创建静态、动态和交互式可视化的库。

## 第2部分：PyTorch核心

### 2.1 Autograd: 自动求导
PyTorch的`autograd`包提供了对Tensors上所有操作的自动微分，是构建神经网络的核心。

```
import torch
print('--- Autograd 示例 ---')

# requires_grad=True 表示需要计算梯度
x = torch.tensor(3.0, requires_grad=True)
y = x**2
print(f'y = x^2 = {y}')

# 计算梯度
y.backward()
print(f'dy/dx at x=3: {x.grad}') # 理论值: 2x = 6

# 高阶导数
z = x**3
grad1 = torch.autograd.grad(z, x, create_graph=True)[0]
grad2 = torch.autograd.grad(grad1, x)[0]
print(f'd(x^3)/dx = 3x^2 = {grad1}')
print(f'd^2(x^3)/dx^2 = 6x = {grad2}')
```

### 2.2 构建神经网络
使用`torch.nn`包来构建、训练和评估神经网络。

```
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
print('--- 神经网络训练流程 ---')

input_size, hidden_size, output_size, batch_size = 10, 32, 2, 64
# 1. 定义模型
model = nn.Sequential(nn.Linear(input_size, hidden_size), nn.ReLU(), nn.Linear(hidden_size, output_size))
# 2. 准备数据
X = torch.randn(100, input_size)
y = torch.randint(0, output_size, (100,))
loader = DataLoader(TensorDataset(X, y), batch_size=batch_size)
# 3. 定义损失和优化器
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
# 4. 训练循环
model.train()
for epoch in range(2):
    for inputs, labels in loader:
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
    print(f'Epoch {epoch+1}, Loss: {loss.item():.4f}')

# 5. 评估
model.eval()
with torch.no_grad():
    correct = (model(X).argmax(dim=1) == y).sum()
    print(f'Accuracy: {correct.item() / len(y) * 100:.2f}%')
```

## 第3部分：高级模型与项目

### 3.1 卷积神经网络 (CNN)
CNNs是图像识别任务的理想选择，它们通过卷积核提取空间特征。

```
print('--- CNN 核心模块 ---')
# 定义一个卷积块
conv_block = nn.Sequential(
    nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, padding=1),
    nn.BatchNorm2d(32),
    nn.ReLU(),
    nn.MaxPool2d(2)
)
dummy_image = torch.randn(1, 3, 32, 32) # (batch, channels, height, width)
output = conv_block(dummy_image)
print(f'输入形状: {dummy_image.shape}')
print(f'输出形状: {output.shape}')
```

### 3.2 循环神经网络 (LSTM/GRU)
LSTMs和GRUs是为处理序列数据（如文本或时间序列）而设计的，它们能有效捕捉长期依赖关系。

```
print('--- LSTM 核心模块 ---')
# 定义一个LSTM层
lstm = nn.LSTM(input_size=20, hidden_size=50, num_layers=2, batch_first=True)
dummy_sequence = torch.randn(1, 10, 20) # (batch, seq_length, feature_size)
output, (hidden, cell) = lstm(dummy_sequence)
print(f'输入序列形状: {dummy_sequence.shape}')
print(f'输出形状: {output.shape}')
print(f'隐藏状态形状: {hidden.shape}')
```
